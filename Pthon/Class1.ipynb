```python
def greatest(a: int, b: int, c: int):
    if a >= b and a >= c:
        print("Greatest:", a)
    elif b >= a and b >= c:
        print("Greatest:", b)
    else:
        print("Greatest:", c)

greatest(2, 3, 4)
```

```python
class Product:
    def __init__(self):
        self.Pr_name = "Prod-1"


class Child(Product):
    def __init__(self):
        super().__init__()
        self.weight = "3"


class Child2(Child):
    def __init__(self):
        super().__init__()
        self.price = "$20"

    def display(self):
        print(self.Pr_name, self.weight, self.price)


obj = Child2()
obj.display()
```
