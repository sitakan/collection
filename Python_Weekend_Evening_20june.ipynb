## While Loop

Runs till a condition is True. If the condition becomes False, STOPS

```python
while condition:
  code
```

```python
#     False
while 10<5:
  print("Hi") # Runs 0 times because the Condition is False
```

```python
#     True
while 10>5:
  print("Hi") # Keeps on running
```

```python
# 1 to 10

# i= 1  -> START
# i= 10 -> END

i=1

while i<11:
  print(i)
  i=i+1

# i=1, 2, 3, 4.....10, 11(STOP)
# Output-> 1, 2, 3,.......9, 10
```

```python
n=int(input("Enter number: "))

i=1
while i<11:
  print(n*i)
  i=i+1
```

```python
# 100,95,90,85....5,0
i=100
# stopping condition -> i>=0

while i>=0:
  print(i,end=" ")
  i-=5
```

```python
correct_password= "Abc@1234"
user_guess=""

# While password incorrect, ask to input.
# Stop when the password is right
while user_guess!=correct_password:
  user_guess=input("Enter password: ")

print("Phone unlocked!")
```

### Control Flow statements.

Used to control the flow of the loops.

1)**break** -> used to stop the loop completely.

2)**continue**-> used to skip a try

```python
# Running a loop from 1 to 10
# STOP when the number is divisible by 7

i=1
while i<=10:
  if i%7==0:
    print("Breaking")
    break
  # Else
  print(i)
  i+=1
```

```python
# continue

# Friends -> A B C D E F G H
#                |     |
#                 Fight


for i in "ABCDEFGH":
  if i=="C" or i=="F":
    print("----Dont give candy to---",i)
    continue # Skip the codes written below when condition is TRUE
  print("Give Chololate to",i)
```

```python
# Find the sum of numbers from 1 to 20.
# Skip when multiple of 7 ( 7 and 14 skipped)
sum=0
for i in range(1,21):
  if i%7==0:
    print("SKIPPING",i)
    continue

  sum=sum+i

print("SUM-",sum)
```

**Return**-> gives a value in return when the function is called.


The moment a return statemet is run, every code below that we ignore, and we come out of the fucntion.

```python
def area(l,b):
  return l*b
```

```python
# cost of land=> area of land X cost per meter sq

AREA = area(13,15) # Storing the l*b in AREA
cost= AREA * 15000
print(cost,"rs")
```

```python
def sample():
  print("ABOVE THE RETURN")
  return -1
  print("BELOW THE RETURN") #Ignored


sample()
```

```python
def largest(l):
  m=l[0]

  for i in l:
    if i>m: # if item is bigger than max
       # updated max to the new item
       m=i

  return m # Return the max element
```

```python
print(largest([9,5,15,12]))
print(largest([-9,55,157,1224,1000]))
```
